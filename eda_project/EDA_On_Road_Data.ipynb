## EDA on Road Accidents in India

### Introduction

Road accidents are a major public safety concern in India, resulting in a significant number of fatalities, injuries, and economic losses each year. With the rapid increase in population, urbanization, and vehicle usage, the need to analyze and understand road accident patterns has become more important than ever.

This project, titled “Analysis of Road Accidents in India,” is based on performing Exploratory Data Analysis (EDA) using the given dataset related to road accidents. The dataset contains relevant information about accidents across different regions, enabling a detailed examination of patterns, trends, and variations in accident occurrences.
The main objective of this project is to explore the dataset through data cleaning, transformation, and visualization techniques in order to extract meaningful insights. EDA helps in identifying key patterns, detecting anomalies, and understanding relationships between different variables such as location, frequency, and other contributing factors.

By applying analytical tools and visualization methods, this study aims to highlight accident-prone areas and understand the overall distribution of road accidents in India. The insights derived from this analysis can support better decision-making, improve road safety measures, and contribute to reducing the risks associated with road transportation.
Overall, this project demonstrates how Exploratory Data Analysis can be effectively used to transform raw accident data into valuable insights, helping to address one of the most critical issues in road safety.

```python
#1. Which top three dates had highest accident in kolkata?
kolkata_df = df[df['City'] == 'Kolkata']
top_dates = kolkata_df['Date'].value_counts().head(3)
print(top_dates)
top_dates.plot(kind='bar')
plt.title("Top 3 Dates with Highest Accidents in Kolkata")
plt.xlabel("Date")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=45)
plt.show()
```

```python
#2. Accidents ocurred in highway due to drunk driving?
highway_drunk = df[(df['Road_type'] == 'highway') & (df['Cause'] == 'drunk driving')]
count = highway_drunk.shape[0]
print("Total Accidents:", count)
highway_drunk['City'].value_counts().head(10).plot(kind='bar')
plt.title("Highway Accidents due to Drunk Driving (Top Cities)")
plt.xlabel("City")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
#3. Which cities had highest accidents?
top_cities = df['City'].value_counts().head(10)
print(top_cities)
top_cities.plot(kind='bar')
plt.title("Top 10 Cities with Highest Accidents")
plt.xlabel("City")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=45)
plt.show()
```

```python
#4. Accidents due to poor road and which cities it happened?
poor_road = df[df['Cause'] == 'poor road']
city_counts = poor_road['City'].value_counts().head(10)
print(city_counts)
city_counts.plot(kind='bar')
plt.title("Accidents due to Poor Road Conditions (Top Cities)")
plt.xlabel("City")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
#5. Categories accidents occured in mumbai and cause of accidents?
mumbai_df = df[df['City'] == 'Mumbai']
category_cause = mumbai_df.groupby(['Accident_severity','Cause']).size().unstack()
print(category_cause)
category_cause.plot(kind='bar', stacked=True)
plt.title("Accident Severity & Causes in Mumbai")
plt.xlabel("Severity")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
#6.Cities which had accident due to weather.
top_cities = df['City'].value_counts().head(5).index
filtered = df[df['City'].isin(top_cities)]
weather_city = filtered.groupby(['City','Weather']).size().unstack()
weather_city.plot(kind='bar')
plt.title("Weather Impact on Accidents in Top Cities")
plt.xlabel("City")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
#7.What are the most common causes of accidents?
cause_counts = df['Cause'].value_counts().head(5)
cause_counts.plot(kind='pie', autopct='%1.1f%%')
plt.title("Most Common Causes of Accidents")
plt.ylabel("")
plt.show()
```

```python
#8. What are the number of accidents ocurred in a hours of a day?
df['Hour'].value_counts().sort_index().plot(kind='line', marker='o')
plt.title("Accidents by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Accidents")
plt.grid()
plt.show()
```

```python
#9. What are the number of vehicles involved vs accidents
df['Vehicles_involved'].value_counts().sort_index().plot(kind='bar')
plt.title("Vehicles Involved in Accidents")
plt.xlabel("Number of Vehicles")
plt.ylabel("Frequency")
plt.show()
```

```python
#10.Which day of the week has the most accidents?
df['Day_of_week'].value_counts().plot(kind='bar')
plt.title("Accidents by Day of Week")
plt.xlabel("Day")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
df['Is_peak_hour'].value_counts().plot(kind='bar')
plt.title("Peak vs Non-Peak Hour Accidents")
plt.xlabel("Peak Hour (1) / Non-Peak (0)")
plt.ylabel("Count")
plt.show()
```
